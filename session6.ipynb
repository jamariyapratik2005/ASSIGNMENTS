# Session 6 - Missing values and data cleaning with pandas

```python
import pandas as pd
import numpy as np
from pathlib import Path

# 1) Load a sample IPL player stats CSV and inspect missing values
base_dir = Path('d:/TOPS/PY_LIBRARY/ASSIGNMENTS/session6_data')
ipl_df = pd.read_csv(base_dir / 'ipl_player_stats.csv')

print('Missing values per column:')
print(ipl_df.isnull().sum())
print('\nNon-missing values per column:')
print(ipl_df.notnull().sum())
```

```python
# 2) Drop rows with any missing values
shape_before = ipl_df.shape
ipl_df_dropped = ipl_df.dropna(axis=0, how='any')
shape_after = ipl_df_dropped.shape

print('Shape before dropping:', shape_before)
print('Shape after dropping:', shape_after)
print(ipl_df_dropped.head())
```

```python
# 3) Fill missing values in the runs column with the mean
ipl_df['runs'] = ipl_df['runs'].fillna(ipl_df['runs'].mean())
print(ipl_df['runs'].head(10))
```

```python
# 4) Simulate a Zomato-style restaurant ratings dataset with missing values
restaurants = pd.DataFrame({
    'restaurant': ['A', 'B', 'C', 'D', 'E'],
    'rating': [4.5, np.nan, 4.2, np.nan, 4.8]
})

print('Before forward/backward fill:')
print(restaurants)

restaurants['rating_ffill'] = restaurants['rating'].ffill()
restaurants['rating_bfill'] = restaurants['rating'].bfill()

print('\nAfter forward fill:')
print(restaurants[['restaurant', 'rating_ffill']])
print('\nAfter backward fill:')
print(restaurants[['restaurant', 'rating_bfill']])
```
