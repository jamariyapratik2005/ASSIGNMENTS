```python
import pandas as pd

# 1) Zomato order history duplicates
zomato_data = pd.DataFrame({
    'restaurant': ['Pizza Hut', 'Pizza Hut', 'Domino\'s', 'KFC', 'Pizza Hut'],
    'order_date': ['2026-08-01', '2026-08-01', '2026-08-02', '2026-08-03', '2026-08-01'],
})

duplicate_orders = zomato_data[zomato_data.duplicated(subset=['restaurant', 'order_date'], keep=False)]
print('Duplicate orders based on restaurant and date:')
print(duplicate_orders)

# 2) Flipkart review duplicates
flipkart_reviews = pd.Series([
    'Great product', 'Amazing quality', 'Great product', 'Good service',
    'Amazing quality', 'Great product', 'Nice packaging', 'Good service'
])

review_counts = flipkart_reviews.value_counts()
duplicate_reviews = review_counts[review_counts > 1]

print('\nTop 3 repeated review texts:')
print(duplicate_reviews.head(3))

# 3) Spotify playlists duplicates
spotify_playlists = pd.DataFrame({
    'playlist_name': ['Chill Mix', 'Workout', 'Chill Mix', 'Study Beats', 'Workout'],
    'creator': ['alex', 'mike', 'alex', 'nina', 'mike']
})

cleaned_playlists = spotify_playlists.drop_duplicates()
print('\nCleaned Spotify playlists:')
print(cleaned_playlists)

# 4) Instagram usernames normalization
instagram_users = pd.DataFrame({
    'username': ['insta_queen', 'insta-queen', 'instaqueen', 'insta_queen', 'dummy_user']
})

instagram_users['username'] = instagram_users['username'].replace({
    'insta_queen': 'instaqueen',
    'insta-queen': 'instaqueen',
    'instaqueen': 'instaqueen'
})
print('\nStandardized Instagram usernames:')
print(instagram_users)

# 5) Paytm-like payment status cleanup
payment_status = pd.Series(['Yes', ' yes ', 'Y', 'No', ' no ', 'N', 'YES', 'n', 'y'])
payment_status = payment_status.str.strip().str.lower()
payment_status = payment_status.map({'yes': 1, 'y': 1, 'no': 0, 'n': 0})
print('\nUnified payment status values:')
print(payment_status)
```
