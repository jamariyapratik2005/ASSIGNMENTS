### Task 2: Zomato Restaurants - Mumbai vs Delhi Comparison (Sweetviz)

**Objective**: Create a visual comparison report between Zomato restaurant data for Mumbai and Delhi using `sweetviz.compare()`.

```python
import pandas as pd
import sweetviz as sv

# Load datasets
df_mumbai = pd.read_csv('session15_data/zomato_mumbai.csv')
df_delhi = pd.read_csv('session15_data/zomato_delhi.csv')

# Generate Sweetviz comparison report
report = sv.compare([df_mumbai, 'Mumbai'], [df_delhi, 'Delhi'])
report.show_html('session15_data/zomato_sweetviz_compare.html', open_browser=False)

print('Comparison Report saved to: session15_data/zomato_sweetviz_compare.html')
print(f"Mumbai Mean Average Cost for Two: ₹{df_mumbai['Average_Cost_for_Two'].mean():.2f}")
print(f"Delhi Mean Average Cost for Two: ₹{df_delhi['Average_Cost_for_Two'].mean():.2f}")
```

**Key Differences Observed in Visual Report**:
1. **Average Cost for Two**: Mumbai restaurants have a significantly higher mean cost for two (₹1,338.67) compared to Delhi (₹784.00).
2. **Online Delivery**: Mumbai shows higher online delivery availability (88%) compared to Delhi (70%).

### Task 4: Flipkart Product Reviews - Report Interpretation & Data Cleaning

**Objective**: Identify two columns in the Flipkart product reviews dataset that require data cleaning or transformation before further analysis.

```python
import pandas as pd

# Load Flipkart reviews dataset
flipkart_df = pd.read_csv('session15_data/flipkart_reviews.csv')

print('=== Dataset Summary ===')
flipkart_df.info()

print('\n1. Sample values of Helpful_Count:')
print(flipkart_df['Helpful_Count'].head(5))

print('\n2. Missing values in Review_Date:')
print(f"{flipkart_df['Review_Date'].isnull().sum()} missing rows out of {len(flipkart_df)}")
```

**Columns Requiring Transformation/Cleaning**:
1. **`Helpful_Count`**: Stored as text strings (`"12 people found this helpful"`). Needs numeric extraction (`str.extract(r'(\d+)')`) and casting to integer.
2. **`Review_Date`**: Contains missing values (`NaN`) and text formatted dates requiring `pd.to_datetime()` parsing and imputation.

### Task 5: Swiggy Food Orders - Sweetviz Summary Report

**Objective**: Generate a summary report for Swiggy food order CSV using Sweetviz.

```python
import pandas as pd
import sweetviz as sv

# Load Swiggy dataset
swiggy_df = pd.read_csv('session15_data/swiggy_orders.csv')

# Generate Sweetviz Summary Report
swiggy_report = sv.analyze(swiggy_df)
swiggy_report.show_html('session15_data/swiggy_sweetviz_report.html', open_browser=False)

print('Swiggy Summary Report saved to: session15_data/swiggy_sweetviz_report.html')
print('\n=== Summary Statistics ===')
print(swiggy_df.describe())
```
