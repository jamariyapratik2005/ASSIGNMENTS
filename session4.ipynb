```python
import pandas as pd
from pathlib import Path

base_dir = Path.cwd()
if (base_dir / "ASSIGNMENTS" / "session4_data").exists():
    base_dir = base_dir / "ASSIGNMENTS" / "session4_data"
elif (base_dir / "session4_data").exists():
    base_dir = base_dir / "session4_data"
else:
    base_dir = Path("session4_data")

# 1. IPL CSV
ipl_df = pd.read_csv(base_dir / "ipl_matches.csv")
print("First 5 IPL rows:")
print(ipl_df.head())

# 2. Trending songs JSON
songs_df = pd.read_json(base_dir / "trending_songs.json")
print("\nTrending songs information:")
songs_df.info()

# 3. Zomato TSV
zomato_df = pd.read_csv(base_dir / "zomato_restaurants.tsv", sep="\t")
print("\nZomato summary:")
print(zomato_df.describe(include="all"))

# 4. Flipkart Excel
flipkart_df = pd.read_excel(base_dir / "flipkart_products.xlsx")

chunk_size = 2000

print("\nFlipkart chunks:")
for start in range(0, len(flipkart_df), chunk_size):
    chunk = flipkart_df.iloc[start:start + chunk_size]
    print("Rows in chunk:", len(chunk))

# 5. Paytm transactions
paytm_df = pd.read_csv(base_dir / "paytm_transactions.csv", sep=";")

missing_values = paytm_df.isnull().sum()

print("\nColumns with null values:")
for column in missing_values.index:
    if missing_values[column] > 0:
        print(column, ":", missing_values[column])
```
