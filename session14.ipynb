```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1) IPL dataset: create a small CSV-like DataFrame and do univariate analysis
ipl_data = pd.DataFrame({
    "match_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "total_runs": [180, 210, 195, 168, 220, 176, 205, 190]
})
summary = ipl_data["total_runs"].agg(["mean", "median", "min", "max", "std"])
print("IPL total_runs summary statistics:")
print(summary)

# 2) Flipkart reviews: bar plot of review counts by rating
flipkart_reviews = pd.DataFrame({
    "rating": [5, 4, 5, 3, 2, 4, 5, 1, 4, 3, 5, 2, 4, 5, 3],
    "category": ["Mobiles", "Fashion", "Electronics", "Home", "Books", "Mobiles", "Fashion", "Electronics",
                 "Home", "Books", "Mobiles", "Fashion", "Electronics", "Home", "Books"]
})
plt.figure(figsize=(7, 4))
sns.countplot(data=flipkart_reviews, x="rating", order=[1, 2, 3, 4, 5], palette="viridis", hue="rating", legend=False)
plt.title("Count of Flipkart Reviews by Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Reviews")
plt.tight_layout()
plt.show()

# 3) Zomato listings: scatterplot between cost and user rating
zomato_data = pd.DataFrame({
    "average_cost_for_two": [500, 800, 1200, 650, 1800, 900, 1500, 1100],
    "user_rating": [4.2, 4.0, 3.8, 4.1, 3.5, 4.3, 3.7, 3.9]
})
plt.figure(figsize=(7, 4))
sns.scatterplot(data=zomato_data, x="average_cost_for_two", y="user_rating", color="crimson")
plt.title("Zomato: Cost vs User Rating")
plt.xlabel("Average Cost for Two")
plt.ylabel("User Rating")
plt.tight_layout()
plt.show()

# 4) BookMyShow movie dataset: groupby analysis by genre
bookmyshow_data = pd.DataFrame({
    "genre": ["Action", "Comedy", "Drama", "Action", "Comedy", "Drama", "Action", "Comedy"],
    "box_office_collection": [120, 85, 90, 130, 95, 88, 125, 92]
})
genre_summary = bookmyshow_data.groupby("genre")["box_office_collection"].mean().sort_values(ascending=False)
print("\nAverage box office collection by genre:")
print(genre_summary)

# 5) Spotify songs: pairplot for selected features
spotify_data = pd.DataFrame({
    "danceability": [0.75, 0.62, 0.84, 0.71, 0.88, 0.55, 0.81, 0.68, 0.79, 0.66],
    "energy": [0.80, 0.70, 0.65, 0.72, 0.90, 0.60, 0.78, 0.69, 0.83, 0.74],
    "valence": [0.70, 0.55, 0.65, 0.60, 0.85, 0.45, 0.72, 0.58, 0.77, 0.62],
    "popularity": [88, 72, 80, 75, 91, 65, 84, 71, 87, 73]
})
pairplot = sns.pairplot(spotify_data)
pairplot.fig.suptitle("Spotify Feature Pairwise Relationships", y=1.02)
plt.show()

print("\nObservation: Higher popularity tends to appear alongside higher danceability and energy in this sample.")
```
