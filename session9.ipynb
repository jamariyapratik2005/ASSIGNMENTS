```python
import pandas as pd
import numpy as np
from pathlib import Path

# 1) Convert string timestamps to datetime
timestamps = ['2024-06-01 14:30', '2024-06-02 09:15', '2024-06-03 20:45']
datetime_series = pd.to_datetime(timestamps)
print('Converted datetime objects:')
print(datetime_series)

# 2) Load the actual order history CSV from the data folder
csv_path = Path('d:/TOPS/PY_LIBRARY/ASSIGNMENTS/session9_data/order_history.csv')
orders_df = pd.read_csv(csv_path, parse_dates=['order_date'])
orders_df['year'] = orders_df['order_date'].dt.year
orders_df['month'] = orders_df['order_date'].dt.month
orders_df['weekday'] = orders_df['order_date'].dt.day_name()
print('\nOrders with extracted date parts from the actual CSV file:')
print(orders_df)

# 3) Set datetime index and resample by week
orders_df = orders_df.set_index('order_date')
weekly_orders = orders_df.resample('W').size()
print('\nOrders per week:')
print(weekly_orders)

# 4) Convert UTC to Asia/Kolkata timezone
instagram_posts = pd.DataFrame({
    'posted_at': [
        '2024-06-01 14:30:00',
        '2024-06-02 09:15:00',
        '2024-06-03 20:45:00',
        '2024-06-04 12:00:00',
        '2024-06-05 18:10:00'
    ]
})
instagram_posts['posted_at'] = pd.to_datetime(instagram_posts['posted_at'], utc=True)
instagram_posts['posted_at_kolkata'] = instagram_posts['posted_at'].dt.tz_convert('Asia/Kolkata')
print('\nFirst 5 converted times (Asia/Kolkata):')
print(instagram_posts[['posted_at_kolkata']].head(5))

# 5) Create weekend feature
orders_df['is_weekend'] = orders_df.index.dayofweek >= 5
print('\nWeekend indicator:')
print(orders_df[['is_weekend']])
```
