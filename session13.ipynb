```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

# 1) Pairplot for the tips dataset
Tips = sns.load_dataset("tips")
numeric_tips = Tips.select_dtypes(include=["number"])
plot1 = sns.pairplot(numeric_tips)
plot1.fig.suptitle("Pairplot of Numeric Variables in Tips Dataset", y=1.02)
plt.show()

# 2) Heatmap for flights dataset
Flights = sns.load_dataset("flights")
flight_pivot = Flights.pivot_table(index="month", columns="year", values="passengers", aggfunc="sum")
flight_pivot = flight_pivot.reindex(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
plt.figure(figsize=(10, 6))
sns.heatmap(flight_pivot, cmap="coolwarm", annot=False)
plt.title("Passengers by Month and Year")
plt.xlabel("Year")
plt.ylabel("Month")
plt.tight_layout()
plt.show()

# 3) Relplot for fmri dataset
Fmri = sns.load_dataset("fmri")
plot3 = sns.relplot(data=Fmri, x="timepoint", y="signal", kind="line", hue="event", marker="o")
plot3.fig.suptitle("Signal Change Over Time by Event Type", y=1.02)
plt.show()

# 4) Catplot for titanic dataset
plot4 = sns.catplot(
    data=Titanic,
    x="class",
    y="survived",
    hue="class",
    kind="bar",
    palette="Set2",
    legend=False
)

plot4.fig.suptitle("Survival Rate by Class", y=1.02)
plt.show()

# 5) Jointplot for penguins dataset
Penguins = sns.load_dataset("penguins")
Penguins = Penguins.dropna(subset=["bill_length_mm", "flipper_length_mm"])
plot5 = sns.jointplot(data=Penguins, x="bill_length_mm", y="flipper_length_mm", kind="reg", color="royalblue")
plot5.fig.suptitle("Relationship Between Bill Length and Flipper Length", y=1.02)
plt.show()

print("Observation: In the penguins sample, longer bills tend to be associated with longer flippers.")
```
