```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_theme(style="whitegrid")

# 1) Histogram of delivery times for 50 Zomato orders
np.random.seed(42)
delivery_times = np.random.normal(loc=35, scale=8, size=50).clip(10, 90)
plt.figure(figsize=(8, 5))
sns.histplot(delivery_times, bins=10, kde=True, color="tomato")
plt.title("Distribution of Zomato Delivery Times")
plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

# 2) Boxplot of total bill by day from the tips dataset
tips = sns.load_dataset("tips")
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="day", y="total_bill", order=["Thur", "Fri", "Sat", "Sun"], palette="Set2", hue="day", legend=False)
plt.title("Total Bill Distribution by Day of the Week")
plt.xlabel("Day")
plt.ylabel("Total Bill ($)")
plt.tight_layout()
plt.show()

# 3) Violinplot comparing IPL team scores
ipl_scores = {
    "MI": [180, 195, 168, 210, 190, 205, 176, 188, 200, 215, 182, 197],
    "CSK": [175, 185, 170, 198, 192, 208, 179, 186, 201, 214, 181, 194],
    "RCB": [190, 202, 178, 220, 205, 212, 188, 199, 208, 224, 196, 210],
    "KKR": [165, 172, 160, 182, 175, 188, 167, 171, 180, 190, 166, 178],
    "SRH": [160, 167, 155, 178, 171, 184, 162, 169, 177, 186, 158, 173],
    "PBKS": [176, 183, 171, 194, 188, 200, 177, 184, 192, 203, 174, 189],
    "RR": [182, 194, 169, 204, 196, 206, 181, 193, 202, 213, 180, 195],
    "DC": [172, 181, 166, 189, 184, 197, 173, 180, 188, 199, 170, 185],
}
ipl_df = pd.DataFrame([(team, score) for team, scores in ipl_scores.items() for score in scores], columns=["Team", "Score"])
ipl_df["Team"] = pd.Categorical(ipl_df["Team"], categories=list(ipl_scores.keys()), ordered=True)

sns.set_theme(style="darkgrid")
plt.figure(figsize=(10, 6))
sns.violinplot(data=ipl_df, x="Team", y="Score", palette="viridis", hue="Team", legend=False)
plt.title("IPL Match Score Distribution by Team")
plt.xlabel("Team")
plt.ylabel("Score")
plt.tight_layout()
plt.show()

# 4) Countplot of Spotify track genres
tracks = [
    "Pop", "Rock", "Hip-Hop", "Jazz", "Pop", "Rock", "Hip-Hop", "Jazz", "Pop", "Rock",
    "Hip-Hop", "Jazz", "Pop", "Rock", "Hip-Hop", "Jazz", "Pop", "Rock", "Hip-Hop", "Jazz",
    "Pop", "Rock", "Hip-Hop", "Jazz", "Pop", "Rock", "Hip-Hop", "Jazz", "Pop", "Rock",
    "Hip-Hop", "Jazz", "Pop", "Rock", "Hip-Hop", "Jazz", "Pop", "Rock", "Hip-Hop", "Jazz"
]
spotify_df = pd.DataFrame({"genre": tracks})
plt.figure(figsize=(8, 5))
sns.countplot(data=spotify_df, x="genre", order=spotify_df["genre"].value_counts().index, palette="magma", hue="genre", legend=False)
plt.title("Number of Spotify Tracks by Genre")
plt.xlabel("Genre")
plt.ylabel("Count")
plt.tight_layout()
plt.show()

# 5) KDE plot of daily step counts for a week
steps = [8200, 10400, 9100, 12300, 7800, 10200, 11800]
steps_df = pd.DataFrame({"steps": steps})

sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))
sns.kdeplot(data=steps_df, x="steps", fill=True, color="seagreen", linewidth=2)
plt.title("Distribution of Daily Step Counts")
plt.xlabel("Steps per Day")
plt.ylabel("Density")
plt.tight_layout()
plt.show()
```
