```python
import matplotlib.pyplot as plt
import numpy as np

# 1) 2x2 grid with four chart types
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Line chart
x = np.arange(1, 6)
y = [3, 5, 4, 6, 7]
axes[0, 0].plot(x, y, marker='o', color='navy')
axes[0, 0].set_title('Line Chart')
axes[0, 0].set_xlabel('X')
axes[0, 0].set_ylabel('Y')

# Bar chart
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 8, 12]
axes[0, 1].bar(categories, values, color='teal')
axes[0, 1].set_title('Bar Chart')
axes[0, 1].set_xlabel('Category')
axes[0, 1].set_ylabel('Value')

# Scatter chart
x_scatter = [1, 2, 3, 4, 5]
y_scatter = [5, 9, 4, 8, 7]
axes[1, 0].scatter(x_scatter, y_scatter, color='crimson')
axes[1, 0].set_title('Scatter Plot')
axes[1, 0].set_xlabel('X')
axes[1, 0].set_ylabel('Y')

# Pie chart
labels = ['Food', 'Travel', 'Shopping', 'Bills']
sizes = [35, 20, 25, 20]
axes[1, 1].pie(sizes, labels=labels, autopct='%1.1f%%', startangle=90)
axes[1, 1].set_title('Pie Chart')

plt.tight_layout()
plt.show()

# 2) Bar chart for delivery times of Zomato, Swiggy, and Domino's
platforms = ['Zomato', 'Swiggy', 'Domino\'s']
avg_delivery_times = [30, 25, 40]
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
linestyles = ['-', '--', ':']
linewidths = [2, 2.5, 3]

plt.figure(figsize=(7, 4))
for i, (platform, time, color, ls, lw) in enumerate(zip(platforms, avg_delivery_times, colors, linestyles, linewidths)):
    plt.bar(platform, time, color=color, linestyle=ls, linewidth=lw, label=platform)
plt.title('Average Delivery Times')
plt.xlabel('Platform')
plt.ylabel('Delivery Time (minutes)')
plt.legend()
plt.tight_layout()
plt.show()

# 3) Multi-axis chart for Instagram followers and average daily posts
influencers = ['A', 'B', 'C', 'D', 'E']
followers = [1200000, 800000, 950000, 1500000, 600000]
posts_per_day = [3.5, 2.2, 4.0, 2.8, 1.9]

fig, ax1 = plt.subplots(figsize=(8, 4.5))
ax2 = ax1.twinx()
ax1.plot(influencers, followers, color='blue', marker='o', label='Followers')
ax1.set_xlabel('Influencer')
ax1.set_ylabel('Followers', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

ax2.plot(influencers, posts_per_day, color='red', marker='s', label='Posts/Day')
ax2.set_ylabel('Average Daily Posts', color='red')
ax2.tick_params(axis='y', labelcolor='red')

plt.title('Instagram Followers vs Average Daily Posts')
plt.tight_layout()
plt.show()

# 4) Scatter plot for Bollywood movie tickets sold vs IMDB ratings
movies = ['Pathaan', 'Jawan', 'Animal', 'Gadar 2', 'Kalki 2898 AD']
rating = [6.8, 7.2, 6.9, 7.5, 7.0]
tickets = [120, 140, 110, 130, 100]

plt.figure(figsize=(8, 4.5))
plt.scatter(rating, tickets, color='darkorange', s=120)
for x, y, label in zip(rating, tickets, movies):
    plt.annotate(label, (x, y), textcoords='offset points', xytext=(0, 8), ha='center')
plt.title('Bollywood Movie Tickets Sold vs IMDb Rating')
plt.xlabel('IMDb Rating')
plt.ylabel('Tickets Sold (in lakhs)')
plt.tight_layout()
plt.show()

# 5) Annotated chart for Flipkart monthly sales
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [15, 22, 19, 30, 27, 35]

plt.figure(figsize=(7, 4))
plt.plot(months, sales, marker='o', color='purple')
plt.title('Flipkart Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales (in crore)')
plt.annotate('Big Billion Days', xy=('Jun', 35), xytext=('Apr', 38),
             arrowprops=dict(arrowstyle='->', color='red'))
plt.tight_layout()
plt.show()
```
