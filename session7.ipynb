```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Detect outliers in Zomato ratings using IQR
zomato_df = pd.DataFrame({
    'restaurant': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'user_rating': [4.2, 4.5, 4.1, 4.8, 4.3, 4.6, 4.9, 2.0]
})

Q1 = zomato_df['user_rating'].quantile(0.25)
Q3 = zomato_df['user_rating'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outlier_indices = zomato_df.index[(zomato_df['user_rating'] < lower_bound) | (zomato_df['user_rating'] > upper_bound)]
print('Zomato outlier indices:', outlier_indices.tolist())

# 2. Boxplot for Swiggy order amounts
swiggy_df = pd.DataFrame({
    'order_amount': [120, 150, 180, 200, 220, 260, 300, 500, 700, 1000]
})

plt.figure(figsize=(6, 4))
plt.boxplot(swiggy_df['order_amount'])
plt.title('Swiggy Order Amount Boxplot')
plt.ylabel('Order Amount')
plt.show()

# 3. Winsorization for Paytm transactions
paytm_df = pd.DataFrame({
    'transaction_amount': [10, 15, 20, 25, 30, 35, 1000, 1200, 1500, 2000]
})

lower = paytm_df['transaction_amount'].quantile(0.05)
upper = paytm_df['transaction_amount'].quantile(0.95)

paytm_df['transaction_amount'] = paytm_df['transaction_amount'].clip(lower=lower, upper=upper)
print('\nWinsorized transaction statistics:')
print(paytm_df['transaction_amount'].describe())

# 4. Convert Flipkart product prices from strings to numeric
flipkart_df = pd.DataFrame({
    'product': ['Phone', 'Headphones', 'Watch', 'Laptop'],
    'price': ['₹1,299', '₹2,499', '₹3,999', '₹89,999']
})

flipkart_df['price'] = flipkart_df['price'].astype(str).str.replace(r'[^0-9]', '', regex=True).astype(int)
print('\nFlipkart numeric prices:')
print(flipkart_df)

# 5. Convert Spotify premium column to boolean
spotify_df = pd.DataFrame({
    'user': ['A', 'B', 'C', 'D'],
    'is_premium': [True, 'True', 1, 'no']
})

spotify_df['is_premium'] = spotify_df['is_premium'].astype(str).str.strip().str.lower().isin(['true', '1', 'yes'])
print('\nSpotify premium boolean column:')
print(spotify_df)
```
