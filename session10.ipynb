```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path

# 1) Line plot for daily steps
steps = [8200, 9400, 7800, 10200, 8900, 11100, 9600]
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']

plt.figure(figsize=(6, 4))
plt.plot(days, steps, marker='o', color='green')
plt.title('Daily Steps Over Last 7 Days')
plt.xlabel('Day')
plt.ylabel('Steps')
plt.tight_layout()
plt.savefig('d:/TOPS/PY_LIBRARY/ASSIGNMENTS/session10_data/steps_lineplot.png')
plt.close()
print('Saved: steps_lineplot.png')

# 2) Scatter plot for Zomato restaurant ratings vs meal price
restaurants = ['Aroma', 'Biryani House', 'Cafe Bliss', 'Dine Hub', 'Eatery 24',
               'Food Point', 'Grill Lane', 'Hunger Stop', 'Indian Express', 'Junction Cafe']
rating = [4.2, 4.5, 4.0, 4.3, 3.9, 4.1, 4.4, 3.8, 4.6, 4.2]
meal_price = [220, 280, 180, 240, 160, 210, 260, 150, 300, 230]

plt.figure(figsize=(7, 4.5))
plt.scatter(rating, meal_price, color='purple')
for i, restaurant in enumerate(restaurants):
    plt.text(rating[i] + 0.01, meal_price[i] + 5, restaurant)
plt.title('Restaurant Ratings vs Average Meal Price')
plt.xlabel('Zomato Rating')
plt.ylabel('Average Meal Price (₹)')
plt.tight_layout()
plt.show()

# 3) Bar chart for orders placed on Swiggy, Zomato, and Domino's
platforms = ['Swiggy', 'Zomato', 'Domino\'s']
orders = [35, 28, 18]
colors = ['skyblue', 'lightgreen', 'salmon']

plt.figure(figsize=(6, 4))
for platform, order, color in zip(platforms, orders, colors):
    plt.bar(platform, order, color=color, label=platform)
plt.title('Orders Placed in the Last Month')
plt.xlabel('Platform')
plt.ylabel('Number of Orders')
plt.legend()
plt.tight_layout()
plt.show()

# 4) Histogram of Spotify listening session durations
session_durations = [25, 40, 35, 50, 30, 45, 60, 55, 32, 42, 38, 48, 27, 33, 52, 41, 37, 29, 46, 34]

plt.figure(figsize=(6, 4))
plt.hist(session_durations, bins=5, color='coral', edgecolor='black')
plt.title('Distribution of Spotify Listening Session Durations')
plt.xlabel('Duration (minutes)')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()

# 5) Two subplots for social media usage
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
screen_time = [90, 110, 85, 120, 95, 130, 100]
posts_liked = [8, 6, 5, 9, 7, 10, 6]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(days, screen_time, marker='o', color='blue')
axes[0].set_title('Daily Instagram Screen Time')
axes[0].set_xlabel('Day')
axes[0].set_ylabel('Screen Time (min)')

axes[1].bar(days, posts_liked, color='orange')
axes[1].set_title('Posts Liked Per Day')
axes[1].set_xlabel('Day')
axes[1].set_ylabel('Number of Posts Liked')

plt.tight_layout()
plt.savefig('d:/TOPS/PY_LIBRARY/ASSIGNMENTS/session10_data/social_media_usage.png')
plt.close()
print('Saved: social_media_usage.png')
```
